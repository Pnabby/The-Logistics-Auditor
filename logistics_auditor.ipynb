```python
#converting date columns
date_cols = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_cols:
    orders[col] = pd.to_datetime(orders[col], errors="coerce")

orders.info()
```

```python
#checking for duplicates in keys
reviews["order_id"].duplicated().sum(), customers["customer_id"].duplicated().sum()
```

```python
dup_reviews = reviews[reviews["order_id"].duplicated(keep=False)].sort_values("order_id")
dup_reviews.head(20)
```

```python
#count number of each duplicate order id
dup_reviews["order_id"].value_counts().head(10)
```

```python
# converting review dates
review_date_cols = ["review_creation_date", "review_answer_timestamp"]

for col in review_date_cols:
    reviews[col] = pd.to_datetime(reviews[col], errors="coerce")

# keeping the latest review per order
cleaned_reviews = (
    reviews.sort_values("review_answer_timestamp")
           .drop_duplicates(subset="order_id", keep="last")
)

# confirm uniqueness
cleaned_reviews["order_id"].duplicated().sum()
```

```python
reviews.shape, cleaned_reviews.shape
```

```python
#merging the tables
master = orders.merge(cleaned_reviews, on="order_id", how="left")
master = master.merge(customers, on="customer_id", how="left")

master.shape
```

```python
#check for duplicates once more
master["order_id"].duplicated().sum()
```

```python
master.head()
```

The "Real" Delay Calculator

```python
# use only the date part because estimate time only 00:00:00
master["delivered_date"] = master["order_delivered_customer_date"].dt.normalize()
master["estimated_date"] = master["order_estimated_delivery_date"].dt.normalize()

# day difference
master["Days_Difference"] = (
    master["estimated_date"] - master["delivered_date"]
).dt.days

master[["order_id", "order_status", "estimated_date", "delivered_date", "Days_Difference", "review_score"]].head()
```

```python
# exclude canceled and unavailable orders
valid_orders = master[~master["order_status"].isin(["canceled", "unavailable"])].copy()

#classification
def classify_delivery(days_diff):
    if pd.isna(days_diff):
        return "Not Delivered"
    elif days_diff >= 0:
        return "On Time"
    elif days_diff >= -5:
        return "Late"
    else:
        return "Super Late"

valid_orders["delivery_status"] = valid_orders["Days_Difference"].apply(classify_delivery)

valid_orders["delivery_status"].value_counts()
```

```python
valid_orders[["estimated_date","delivered_date","Days_Difference", "delivery_status"]].sample(10)
```

The Geographic Heatmap

```python
#% of late orders per state

state_perf = valid_orders.groupby("customer_state").agg(
    total_orders=("order_id", "count"),
    late_orders=("delivery_status", lambda x: x.isin(["Late", "Super Late"]).sum())
).reset_index()

state_perf["late_percentage"] = (state_perf["late_orders"] / state_perf["total_orders"]) * 100

state_perf.sort_values("late_percentage", ascending=False).head(10)
```

```python
#bar chart
top_late_states = state_perf.sort_values("late_percentage", ascending=False).head(10)

ax = top_late_states.plot(
    kind="bar",
    x="customer_state",
    y="late_percentage",
    figsize=(10, 6),
    title="Top 10 States by Percentage of Late Deliveries",
    legend=False
)

ax.set_xlabel("State")
ax.set_ylabel("Late Delivery Rate (%)")
ax.tick_params(axis="x", rotation=0)

for container in ax.containers:
    ax.bar_label(container, fmt="%.1f")
```

Sentiment Correlation

```python
#Sentiment Correlation

#grouping delivery statuses
review_by_status = valid_orders[
    valid_orders["delivery_status"] != "Not Delivered"
].groupby("delivery_status").agg(
    avg_review_score=("review_score", "mean"),
    count_orders=("order_id", "count")
).reset_index()

review_by_status
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(data=review_by_status, x="delivery_status", y="avg_review_score")
plt.title("Average Review Score by Delivery Status")
plt.show()
```

```python
#getting min and max day difference for bins
delivered = valid_orders[
    valid_orders["delivery_status"] != "Not Delivered"
]

delivered["Days_Difference"].min(), delivered["Days_Difference"].max()
```

```python
# cap to get rid of outliers
delivered["Days_Difference_capped"] = delivered["Days_Difference"].clip(-30, 30)

delivered["delay_bin"] = pd.cut(
    delivered["Days_Difference_capped"],
    bins=[-30, -10, -5, 0, 3, 7, 30],
    labels=[
        "Very Late (>10 days)",
        "Late (5–10 days)",
        "Slightly Late (0–5 days)",
        "On Time",
        "Early (1–3 days)",
        "Very Early (>3 days)"
    ]
)

delay_review = delivered.groupby("delay_bin")["review_score"].mean().reset_index()

sns.barplot(data=delay_review, x="delay_bin", y="review_score")
plt.xticks(rotation=45)
plt.title("Review Score by Delay Range")
plt.show()
```

#### Insight

Customer satisfaction and delivery delays are clearly negatively correlated. The average review score is higher for orders that are delivered on time than for those that are delivered late, especially very late. This confirms that delays in logistics directly impact customer experience.

The "Translation" Challenge

```python

products = pd.read_csv("dataset/olist_products_dataset.csv")
translations = pd.read_csv("dataset/product_category_name_translation.csv")

display(products.head())
display(translations.head())
```

```python
order_items = pd.read_csv("dataset/olist_order_items_dataset.csv")
display (order_items.head())
```

```python
#merging
master_with_products = master.merge(
    order_items[["order_id", "product_id"]],
    on="order_id",
    how="left"
).merge(
    products[["product_id", "product_category_name"]],
    on="product_id",
    how="left"
).merge(
    translations,
    on="product_category_name",
    how="left"
)

master_with_products.head()
```

```python
#testing translation
master_with_products[[
    "product_category_name",
    "product_category_name_english"
]].drop_duplicates().head(20)
```

```python
#translation data
data_translated = master_with_products[
    ~master_with_products["order_status"].isin(["canceled", "unavailable"])
].copy()

category_summary = data_translated.groupby("product_category_name_english").agg(
    total_orders=("order_id", "nunique"),
    avg_review_score=("review_score", "mean"),
    avg_days_difference=("Days_Difference", "mean")
).reset_index()

category_summary.head(30)
```

```python
#trying to get categories with volume
category_summary = category_summary[category_summary["total_orders"] >= 70]
```

```python
print("Worst Delivery timing")
display(category_summary.sort_values("avg_days_difference").head(10))
print("lowest reviews")
display(category_summary.sort_values("avg_review_score").head(10))
```

```python
top_delayed_categories = category_summary.sort_values("avg_days_difference").head(10)

plt.figure(figsize=(10,6))
plt.barh(top_delayed_categories["product_category_name_english"], top_delayed_categories["avg_days_difference"])
plt.xlabel("Average Days Difference")
plt.ylabel("Product Category")
plt.title("Top 10 Product Categories with Worst Delivery Performance")
plt.gca().invert_yaxis()
plt.show()
```

### Insight
Furniture and home-related categories tend to receive lower review scores and also show relatively weaker delivery performance. This suggests that bulky or complex-to-ship products are more likely to result in poor customer experiences.

```python
#My choice
order_items = pd.read_csv("dataset/olist_order_items_dataset.csv")

order_value = order_items.groupby("order_id").agg(
    total_price=("price", "sum"),
    total_freight=("freight_value", "sum")
).reset_index()

order_value.head()
```

```python
master_value = master.merge(order_value, on="order_id", how="left")
#adding a bad review field
master_value["bad_review"] = master_value["review_score"].apply(lambda x: 1 if x <= 2 else 0)
```

```python
state_bad_review = master_value.groupby("customer_state").agg(
    total_orders=("order_id", "nunique"),
    bad_review_orders=("bad_review", "sum"),
    total_revenue=("total_price", "sum"),
    bad_review_revenue=("total_price", lambda x: x[master_value.loc[x.index, "bad_review"] == 1].sum()),
    avg_review_score=("review_score", "mean")
).reset_index()

#bad review percentage
state_bad_review["bad_review_percentage"] = (
    state_bad_review["bad_review_orders"] / state_bad_review["total_orders"]
) * 100

state_bad_review.head()
```

```python
# top 10 states by bad review revenue

top_10_bad_revenue = state_bad_review.sort_values(
    "bad_review_revenue", ascending=False
).head(10)

plt.figure(figsize=(10, 6))
bars = plt.bar(top_10_bad_revenue["customer_state"], top_10_bad_revenue["bad_review_revenue"])

plt.title("Top 10 States by Bad Review Revenue")
plt.xlabel("Customer State")
plt.ylabel("Bad Review Revenue")
plt.xticks(rotation=45)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f"{height:,.0f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

**Insight:** The chart shows that bad review revenue is heavily concentrated in a few key states, especially **SP**, which is far ahead of all others, followed by **RJ** and **MG**. This means that poor customer experiences in these high-revenue states may have a greater business impact than similar issues in lower-volume regions. For Veridi Logistics, these states should be treated as priority areas for operational improvement, since reducing service failures there could help protect customer trust and a significant share of revenue.
